```python
import pandas as pd
import glob

path = "/kaggle/input/datasets/asrarfarooq11/cyclistic-bike-share-trip-data-4-2020-to-3-2021/"

files = glob.glob(path + "*.csv")

print("Number of CSV files:", len(files))
print(files)
```

```python
dfs = []

for file in files:
    temp = pd.read_csv(file)
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)

print(df.shape)
```

```python
df["started_at"] = pd.to_datetime(df["started_at"])
df["ended_at"] = pd.to_datetime(df["ended_at"])
```

```python
df[["started_at", "ended_at"]].dtypes
```

```python
df["ride_length"] = df["ended_at"] - df["started_at"]

df["ride_minutes"] = df["ride_length"].dt.total_seconds() / 60
```

```python
df["ride_minutes"].describe()
```

```python
(df["ride_minutes"] <= 0).sum()
```

```python
(df["ride_minutes"] > 1440).sum()
```

```python
df = df[(df["ride_minutes"] > 0) & (df["ride_minutes"] <= 1440)].copy()

print(df.shape)
```

```python
df["day_of_week"] = df["started_at"].dt.day_name()
df["month"] = df["started_at"].dt.month_name()
df["hour"] = df["started_at"].dt.hour
```

```python
df.groupby("member_casual")["ride_minutes"].agg(["count", "mean", "median"])
```

```python
day_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

day_analysis = df.groupby(
    ["day_of_week", "member_casual"]
).size().unstack()

day_analysis = day_analysis.reindex(day_order)

day_analysis
```

```python
day_analysis.plot(kind="bar", figsize=(10,5))
```

```python
month_order = [
    "April", "May", "June", "July", "August", "September",
    "October", "November", "December", "January", "February", "March"
]

month_analysis = df.groupby(
    ["month", "member_casual"]
).size().unstack()

month_analysis = month_analysis.reindex(month_order)

month_analysis
```

```python
month_analysis.plot(kind="bar", figsize=(12,5))
```

```python
bike_analysis = df.groupby(
    ["rideable_type", "member_casual"]
).size().unstack()

bike_analysis
```

```python
bike_analysis.plot(kind="bar", figsize=(10,5))
```

```python
df["is_weekend"] = df["day_of_week"].isin(["Saturday", "Sunday"])

weekend_analysis = df.groupby(
    ["is_weekend", "member_casual"]
)["ride_minutes"].agg(["count", "mean"])

weekend_analysis
```

```python
weekend_counts = df.groupby(
    ["is_weekend", "member_casual"]
).size().unstack()

weekend_counts
```

```python
weekend_share = df.groupby(
    ["member_casual", "is_weekend"]
).size().unstack()

weekend_share["weekend_percentage"] = (
    weekend_share[True] /
    (weekend_share[True] + weekend_share[False]) * 100
)

weekend_share
```

```python
avg_duration = df.groupby("member_casual")["ride_minutes"].mean()

avg_duration.plot(
    kind="bar",
    figsize=(8,5),
    title="Average Ride Duration: Casual vs Members",
    ylabel="Average Ride Duration (minutes)",
    xlabel="Rider Type"
)
```

```python
weekend_share["weekend_percentage"].plot(
    kind="bar",
    figsize=(8,5),
    title="Percentage of Rides Taken on Weekends",
    ylabel="Weekend Rides (%)",
    xlabel="Rider Type"
)
```
